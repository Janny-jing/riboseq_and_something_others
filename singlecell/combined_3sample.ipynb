```python
import scanpy as sc
import pandas as pd
import scipy.io
import anndata
from scipy.sparse import csr_matrix
import os
import scanpy as sc
import pandas as pd
import numpy as np
import os
import glob
import matplotlib.pyplot as plt
import re
```

```python
def read_anndata_C4(path):   
    mat = scipy.io.mmread(os.path.join(path, "matrix.mtx.gz")).astype("float32")
    mat = mat.transpose()
    mat = csr_matrix(mat)
    adata = anndata.AnnData(mat,dtype="float32")
    genes = pd.read_csv(os.path.join(path, "features.tsv.gz"), header=None, sep='\t')
    var_names = genes[0].values
    var_names = anndata.utils.make_index_unique(pd.Index(var_names))
    adata.var_names = var_names
    adata.var['gene_symbols'] = genes[0].values
    adata.obs_names = pd.read_csv(os.path.join(path, "barcodes.tsv.gz"), header=None)[0].values
    adata.var_names_make_unique()
    return adata
# 自动发现所有样本
def discover_samples(base_path="/storage2/liuxiaodongLab/jiangjing/Projects/XueyingFan/PD_XueyingFan/20251224_bgi_sisth_dataset/output"):
    """发现指定路径下的所有样本目录"""
    sample_dirs = []
    sample_names = []
    
    print(f"在基础路径下查找样本: {base_path}")
    
    # 查找符合模式 C*_sample/C*_sample/output/filter_matrix 的目录
    pattern = os.path.join(base_path, "*_sample/*_sample/output/filter_matrix")
    matches = glob.glob(pattern)
    
    print(f"找到匹配项: {matches}")
    
    for match in matches:
        # 从路径中提取样本名，例如从 ".../C1_sample/C1_sample/..." 中提取 "C1"
        path_parts = match.split('/')
        for part in path_parts:
            if re.match(r'^[A-Za-z]+\d+_sample$', part):  # 匹配 C1_sample, C2_sample 等
                sample_name = part.replace('_sample', '')
                if sample_name not in sample_names:
                    sample_dirs.append(match)
                    sample_names.append(sample_name)
                    print(f"发现样本: {sample_name} -> {match}")
                break
    
    # 如果上面的模式没找到，尝试更简单的模式
    if not sample_dirs:
        print("尝试备用查找模式...")
        alternative_patterns = [
            os.path.join(base_path, "*_sample/output/filter_matrix"),
            os.path.join(base_path, "*/output/filter_matrix"),
        ]
        
        for pattern in alternative_patterns:
            matches = glob.glob(pattern)
            for match in matches:
                sample_name = os.path.basename(os.path.dirname(os.path.dirname(match))).replace('_sample', '')
                if sample_name not in sample_names:
                    sample_dirs.append(match)
                    sample_names.append(sample_name)
                    print(f"发现样本: {sample_name} -> {match}")
    
    return sample_dirs, sample_names

# 读取所有样本数据
def load_all_samples(sample_dirs, sample_names):
    """加载所有样本数据"""
    adata_dict = {}
    for sample_dir, sample_name in zip(sample_dirs, sample_names):
        try:
            print(f"正在读取样本 {sample_name}...")
            adata = read_anndata_C4(sample_dir)
            if adata is not None:
                adata.obs['sample'] = sample_name
                adata.obs_names_make_unique()
                adata_dict[sample_name] = adata
                print(f"成功读取样本 {sample_name}: {adata.shape} (细胞数: {adata.n_obs}, 基因数: {adata.n_vars})")
            else:
                print(f"读取样本 {sample_name} 失败")
        except Exception as e:
            print(f"读取样本 {sample_name} 时出错: {e}")
    
    return adata_dict

# 计算质控指标
def calculate_qc_metrics(adata_dict):
    """为所有样本计算质控指标"""
    for sample_name, adata in adata_dict.items():
        try:
            # 标记线粒体基因（兼容不同命名方式）
            mt_patterns = ['MT-', 'mt-', 'Mt-', '^MT', '^mt']
            mt_genes_found = False
            
            for pattern in mt_patterns:
                adata.var["mt"] = adata.var_names.str.startswith(pattern)
                if adata.var["mt"].sum() > 0:
                    mt_genes_found = True
                    print(f"样本 {sample_name} 使用模式 '{pattern}' 找到 {adata.var['mt'].sum()} 个线粒体基因")
                    break
            
            if not mt_genes_found:
                print(f"警告: 样本 {sample_name} 未找到线粒体基因")
                adata.var["mt"] = False
            
            # 计算质控指标
            sc.pp.calculate_qc_metrics(adata, qc_vars=["mt"], inplace=True, log1p=True)
            print(f"样本 {sample_name} 质控指标计算完成")
            
        except Exception as e:
            print(f"计算样本 {sample_name} 质控指标时出错: {e}")

# 绘制质控小提琴图
def plot_qc_violin(adata_dict, output_dir="qc_plots"):
    """为所有样本绘制质控小提琴图"""
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # 合并所有数据用于绘图
    adata_list = list(adata_dict.values())
    if not adata_list:
        print("没有数据可绘制！")
        return None
    
    adata_combined = sc.concat(adata_list, label='sample', keys=list(adata_dict.keys()))
    
    # 创建更大的图以适应多个样本
    n_samples = len(adata_dict)
    fig_width = max(15, n_samples * 2)  # 根据样本数量调整宽度
    
    fig, axes = plt.subplots(1, 3, figsize=(fig_width, 6))
    
    # n_genes_by_counts
    sc.pl.violin(adata_combined, 'n_genes_by_counts', groupby='sample', ax=axes[0], show=False)
    axes[0].set_title('Number of genes by counts')
    axes[0].tick_params(axis='x', rotation=45)
    
    # total_counts
    sc.pl.violin(adata_combined, 'total_counts', groupby='sample', ax=axes[1], show=False)
    axes[1].set_title('Total counts per cell')
    axes[1].tick_params(axis='x', rotation=45)
    
    # pct_counts_mt
    sc.pl.violin(adata_combined, 'pct_counts_mt', groupby='sample', ax=axes[2], show=False)
    axes[2].set_title('Percentage of mitochondrial counts')
    axes[2].tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.savefig(f"{output_dir}/qc_violin_plots.png", dpi=300, bbox_inches='tight')
    plt.show()
    
    # 保存详细的统计信息
    qc_stats = []
    for sample_name, adata in adata_dict.items():
        stats = {
            'sample': sample_name,
            'n_cells': adata.n_obs,
            'n_genes': adata.n_vars,
            'mean_genes': adata.obs['n_genes_by_counts'].mean(),
            'median_genes': adata.obs['n_genes_by_counts'].median(),
            'std_genes': adata.obs['n_genes_by_counts'].std(),
            'min_genes': adata.obs['n_genes_by_counts'].min(),
            'max_genes': adata.obs['n_genes_by_counts'].max(),
            'mean_counts': adata.obs['total_counts'].mean(),
            'median_counts': adata.obs['total_counts'].median(),
            'std_counts': adata.obs['total_counts'].std(),
            'min_counts': adata.obs['total_counts'].min(),
            'max_counts': adata.obs['total_counts'].max(),
            'mean_pct_mt': adata.obs['pct_counts_mt'].mean(),
            'median_pct_mt': adata.obs['pct_counts_mt'].median(),
            'std_pct_mt': adata.obs['pct_counts_mt'].std(),
            'max_pct_mt': adata.obs['pct_counts_mt'].max()
        }
        qc_stats.append(stats)
    
    qc_df = pd.DataFrame(qc_stats)
    qc_df.to_csv(f"{output_dir}/qc_statistics.csv", index=False)
    print("质控统计信息已保存到 qc_plots/qc_statistics.csv")
    
    # 打印简要统计信息
    print("\n样本统计摘要:")
    print(qc_df[['sample', 'n_cells', 'mean_genes', 'mean_counts', 'mean_pct_mt']].to_string(index=False))
    
    return qc_df

# 主流程
def main():
    # 1. 发现样本
    print("正在发现样本...")
    base_path = "/storage2/liuxiaodongLab/jiangjing/Projects/XueyingFan/PD_XueyingFan/20251224_bgi_sisth_dataset/output"
    sample_dirs, sample_names = discover_samples(base_path)
    
    if not sample_dirs:
        print("未发现任何样本！请检查路径或手动指定样本。")
        return
    
    print(f"\n发现 {len(sample_dirs)} 个样本: {sample_names}")
    
    # 2. 读取样本
    adata_dict = load_all_samples(sample_dirs, sample_names)
    
    if not adata_dict:
        print("没有成功读取任何样本数据！")
        return
    
    # 3. 计算质控指标
    calculate_qc_metrics(adata_dict)
    
    # 4. 绘制质控小提琴图
    print("正在绘制质控小提琴图...")
    qc_stats = plot_qc_violin(adata_dict)
    
    print("\n" + "="*60)
    print("质控小提琴图已生成！请查看 qc_plots/qc_violin_plots.png")
    print("详细质控统计信息已保存到 qc_plots/qc_statistics.csv")
    print("\n请根据小提琴图手动决定过滤阈值，然后继续下一步分析。")
    print("="*60)
    
    return adata_dict

# 运行主流程
if __name__ == "__main__":
    adata_dict = main()
```

```python
def continue_analysis(adata_dict, filter_params):
    """
    根据手动决定的过滤参数继续分析
    """
    
    print("开始质控过滤...")
    print(f"使用统一过滤参数: {filter_params}")
    
    # 质控过滤 - 所有样本使用相同的参数
    filtered_adata_dict = {}
    for sample_name, adata in adata_dict.items():
        initial_cells = adata.n_obs
        filtered_adata = adata[
            (adata.obs['n_genes_by_counts'] > filter_params['min_genes']) & 
            (adata.obs['n_genes_by_counts'] < filter_params['max_genes']) & 
            (adata.obs['total_counts'] > filter_params['min_counts']) & 
            (adata.obs['total_counts'] < filter_params['max_counts']) & 
            (adata.obs['pct_counts_mt'] < filter_params['max_mt'])
        ].copy()
        filtered_adata_dict[sample_name] = filtered_adata
        removed_cells = initial_cells - filtered_adata.n_obs
        removal_rate = (removed_cells / initial_cells) * 100
        print(f"样本 {sample_name}: {initial_cells} -> {filtered_adata.n_obs} 细胞 (移除 {removed_cells} 个细胞, {removal_rate:.1f}%)")
    
    # 后续处理步骤
    sample_list = []
    for sample_name, adata in filtered_adata_dict.items():
        try:
            print(f"处理样本 {sample_name}...")
            
            # 过滤基因和细胞
            initial_genes = adata.n_vars
            sc.pp.filter_genes(adata, min_cells=3)
            after_gene_filter = adata.n_vars
            removed_genes = initial_genes - after_gene_filter
            print(f"  基因过滤: {initial_genes} -> {after_gene_filter} (移除 {removed_genes} 个基因)")
            
            initial_cells = adata.n_obs
            sc.pp.filter_cells(adata, min_genes=200)
            after_cell_filter = adata.n_obs
            removed_cells = initial_cells - after_cell_filter
            print(f"  细胞过滤: {initial_cells} -> {after_cell_filter} (移除 {removed_cells} 个细胞)")
            
            # 运行scrublet检测双细胞
            print("  运行Scrublet检测双细胞...")
            sc.external.pp.scrublet(adata)
            
            # 去除双细胞
            if 'predicted_doublet' in adata.obs.columns:
                initial_cells = adata.n_obs
                doublet_count = adata.obs['predicted_doublet'].sum()
                adata = adata[~adata.obs['predicted_doublet']].copy()
                removed_doublets = initial_cells - adata.n_obs
                print(f"  双细胞移除: {initial_cells} -> {adata.n_obs} (移除 {removed_doublets} 个双细胞, 检测到 {doublet_count} 个)")
            
            # 关键修改：在合并前为每个样本保存原始counts
            adata.layers["counts"] = adata.X.copy()
            
            sample_list.append(adata)
            print(f"  样本 {sample_name} 处理完成: {adata.shape}\n")
            
        except Exception as e:
            print(f"处理样本 {sample_name} 时出错: {e}")
    
    # 整合所有样本
    if sample_list:
        print("整合所有样本...")
        adata_combined = sc.concat(sample_list, 
                                  join='outer',
                                  label='sample', 
                                  keys=list(filtered_adata_dict.keys()))
        
        # 关键修改：确保counts层包含原始数据
        # 如果合并后counts层丢失，从各个样本中重建
        if 'counts' not in adata_combined.layers:
            print("重建counts层...")
            # 创建一个空的counts矩阵
            counts_matrix = adata_combined.X.copy()
            
            # 用原始数据填充counts矩阵
            start_idx = 0
            for adata in sample_list:
                if 'counts' in adata.layers:
                    end_idx = start_idx + adata.n_obs
                    counts_matrix[start_idx:end_idx] = adata.layers['counts']
                    start_idx = end_idx
            
            adata_combined.layers["counts"] = counts_matrix
        
        # 标准化和log转换
        print("进行数据标准化...")
        sc.pp.normalize_total(adata_combined, target_sum=10000)
        sc.pp.log1p(adata_combined)
        adata_combined.layers["logcounts"] = adata_combined.X.copy()
        
        # # 选择高变基因
        # print("选择高变基因...")
        # sc.pp.highly_variable_genes(adata_combined, n_top_genes=2000, batch_key='sample')
        
        # # 基础分析流程
        # print("进行PCA降维...")
        # sc.tl.pca(adata_combined)
        
        # print("计算邻居图...")
        # sc.pp.neighbors(adata_combined, n_pcs=30)
        
        # print("进行UMAP降维...")
        # sc.tl.umap(adata_combined)
        
        # print("进行Leiden聚类...")
        # sc.tl.leiden(adata_combined)
        
        # # 验证counts层数据
        # print("验证数据层:")
        # print(f"  counts层总和: {adata_combined.layers['counts'].sum():.0f}")
        # print(f"  logcounts层总和: {adata_combined.layers['logcounts'].sum():.2f}")
        # print(f"  X矩阵总和: {adata_combined.X.sum():.2f}")
        
        # print(f"最终整合数据: {adata_combined.shape}")
        # print(f"样本分布: {adata_combined.obs['sample'].value_counts().to_dict()}")
        
        return adata_combined
    else:
        print("没有样本通过过滤！")
        return None
```

```python
filter_params = {
    'min_genes': 2000, 
    'max_genes': 12000, 
    'min_counts': 2000, 
    'max_counts': 150000, 
    'max_mt': 10
}

# 继续分析 - 这是实际执行函数的调用
print("=" * 50)
print("开始执行单细胞数据分析流程")
print("=" * 50)

adata_final = continue_analysis(adata_dict, filter_params)
```

```python
# 绘制每个基因的UMAP图
import matplotlib.pyplot as plt

# 为每个细胞类型分组创建UMAP图
for cell_type, genes in gene_celltype_mapping.items():
    # 过滤出在当前数据中存在的基因
    available_genes_in_group = [gene for gene in genes if gene in com_adata.var_names]
    
    if available_genes_in_group:  # 如果该分组有可用的基因
        # 创建子图
        n_genes = len(available_genes_in_group)
        n_cols = min(4, n_genes)  # 每行最多4个图
        n_rows = (n_genes + n_cols - 1) // n_cols
        
        fig, axes = plt.subplots(n_rows, n_cols, figsize=(4*n_cols, 4*n_rows))
        fig.suptitle(f'{cell_type} - Gene Expression in UMAP', fontsize=16, y=0.95)
        
        # 如果只有一行，确保axes是列表
        if n_rows == 1:
            axes = [axes] if n_cols == 1 else axes
        else:
            axes = axes.flatten()
        
        # 绘制每个基因的UMAP
        for i, gene in enumerate(available_genes_in_group):
            if i < len(axes):
                sc.pl.umap(com_adata, color=gene, ax=axes[i], show=False, 
                          title=gene, cmap='Reds', frameon=False)
        
        # 隐藏多余的子图
        for i in range(len(available_genes_in_group), len(axes)):
            axes[i].axis('off')
        
        plt.tight_layout()
        plt.savefig(f'umap_{cell_type}.pdf', bbox_inches='tight', dpi=300)
        plt.show()

# 或者绘制所有基因在一个大图中（如果基因数量不多）
all_genes = []
for genes in gene_celltype_mapping.values():
    all_genes.extend(genes)

available_all_genes = [gene for gene in all_genes if gene in com_adata.var_names]

# 绘制所有基因的UMAP
n_genes_total = len(available_all_genes)
n_cols = 5  # 每行5个图
n_rows = (n_genes_total + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(5*n_cols, 5*n_rows))
fig.suptitle('All Genes Expression in UMAP', fontsize=20, y=0.95)

if n_rows == 1:
    axes = [axes] if n_cols == 1 else axes
else:
    axes = axes.flatten()

for i, gene in enumerate(available_all_genes):
    if i < len(axes):
        sc.pl.umap(com_adata, color=gene, ax=axes[i], show=False, 
                  title=gene, cmap='Reds', frameon=False, size=20)

# 隐藏多余的子图
for i in range(len(available_all_genes), len(axes)):
    axes[i].axis('off')

plt.tight_layout()
plt.savefig('umap_all_genes.pdf', bbox_inches='tight', dpi=300)
plt.show()
```
